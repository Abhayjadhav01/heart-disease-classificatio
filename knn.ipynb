```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv(r"heart.csv")
df.head()
```

```python
df.info()
```

```python
x=df.drop("target", axis=1)
y=df["target"]

x_train, x_test, y_train, y_test=train_test_split(
    x, y, test_size=0.2, random_state=42
)

scaler=StandardScaler()
x_train_scaler=scaler.fit_transform(x_train)
x_test_scaler=scaler.transform(x_test)
```

```python
knn_classifier=KNeighborsClassifier(n_neighbors=3)
knn_classifier.fit(x_train_scaler, y_train)

y_pred=knn_classifier.predict(x_test_scaler)
print(y_pred)
```

```python


acc_score=accuracy_score(y_test, y_pred)
print("acc_score = ",acc_score)

precision=precision_score(y_test, y_pred)
print("precision = ",precision)

recall=recall_score(y_test, y_pred)
print("recall = ",recall)

from sklearn.metrics import confusion_matrix
cm_matrix=confusion_matrix(y_test, y_pred)
print(cm_matrix)
```

```python
#K=5

knn_classifier=KNeighborsClassifier(n_neighbors=5)
knn_classifier.fit(x_train_scaler, y_train)

y_pred=knn_classifier.predict(x_test_scaler)



acc_score=accuracy_score(y_test, y_pred)
print("acc_score = ",acc_score)

precision=precision_score(y_test, y_pred)
print("precision = ",precision)

recall=recall_score(y_test, y_pred)
print("recall = ",recall)


from sklearn.metrics import confusion_matrix
cm_matrix=confusion_matrix(y_test, y_pred)
print(cm_matrix)
```

```python
#K=7

knn_classifier=KNeighborsClassifier(n_neighbors=7)
knn_classifier.fit(x_train_scaler, y_train)

y_pred=knn_classifier.predict(x_test_scaler)



acc_score=accuracy_score(y_test, y_pred)
print("acc_score = ",acc_score)

precision=precision_score(y_test, y_pred)
print("precision = ",precision)

recall=recall_score(y_test, y_pred)
print("recall = ",recall)


from sklearn.metrics import confusion_matrix
cm_matrix=confusion_matrix(y_test, y_pred)
print(cm_matrix)
```

```python
classifier=KNeighborsClassifier()
param_grid={"n_neighbors" :[2,3, 5, 7,9]}

from sklearn.model_selection import GridSearchCV
classifiercv=GridSearchCV(
    classifier,
    param_grid,
    cv=5,
    scoring="recall"
)

classifiercv.fit(x_train_scaler, y_train)

y_pred=classifiercv.predict(x_test_scaler)

print("recall = ", recall_score(y_test, y_pred))
print("accuracy = ", accuracy_score(y_test, y_pred))
print("precision = ", precision_score(y_test, y_pred))

res = pd.DataFrame(classifiercv.cv_results_)
print(res[["param_n_neighbors", "mean_test_score"]])
print(classifiercv.best_params_)
```

```python
from sklearn.pipeline import Pipeline

x_train, x_test, y_train, y_test=train_test_split(
    x, y, test_size=0.2, random_state=42
)

pipeline=Pipeline([('scaler', StandardScaler()), ('knn', KNeighborsClassifier())])

param_grid={"knn__n_neighbors" :[2,3, 5, 7,9]}

classifiercv=GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="recall"
)

classifiercv.fit(x_train, y_train)

y_pred=classifiercv.predict(x_test)

print("recall = ", recall_score(y_test, y_pred))
print("accuracy = ", accuracy_score(y_test, y_pred))
print("precision = ", precision_score(y_test, y_pred))
```
